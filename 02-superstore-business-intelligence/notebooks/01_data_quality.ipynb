# SUPERSTORE BUSINESS INTELLIGENCE
 DATA QUALITY CHECK

```python
import pandas as pd
import numpy as np
from pathlib import Path
```

## 1. CONFIGURATION

```python
DATA_PATH = Path("../data/SampleSuperstore.csv")
```

## 2. CHARGEMENT

```python
df = pd.read_csv(DATA_PATH, encoding="latin1")

print("SUPERSTORE - DATA QUALITY")

print(f"\nNombre de lignes : {df.shape[0]:,}")
print(f"Nombre de colonnes : {df.shape[1]}")
```

## 3. APERÇU

```python
print("\n--- Aperçu ---")
print(df.head())
```

## 4. TYPES DE DONNÉES

```python
print("\n--- Types de données ---")
print(df.dtypes)
```

## 5. VALEURS MANQUANTES

```python
print("\n--- Valeurs manquantes ---")

missing = (
    df.isnull()
    .sum()
    .sort_values(ascending=False)
)

print(missing[missing > 0])
```

## 6. DOUBLONS

```python

duplicates = df.duplicated().sum()

print("\nNombre de doublons :", duplicates)
```

## 7. VALEURS UNIQUES

```python
print("\n--- Valeurs uniques ---")

for column in df.columns:

    print(
        f"{column}: "
        f"{df[column].nunique():,} valeurs uniques"
    )
```

## 8. VARIABLES NUMÉRIQUES

```python
numeric_columns = [
    "Sales",
    "Quantity",
    "Discount",
    "Profit"
]

print("\n--- Statistiques numériques ---")

print(
    df[numeric_columns]
    .describe()
    .T
)
```

```python
## 9. VALEURS NÉGATIVES
```

```python
print("\n--- Valeurs négatives ---")

for column in numeric_columns:

    negative_count = (
        df[column] < 0
    ).sum()

    print(
        f"{column}: {negative_count:,}"
    )
```

## 10. CONTRÔLE DES DATES

```python
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    errors="coerce"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    errors="coerce"
)

print("\nDates invalides :")

print(
    "Order Date :",
    df["Order Date"].isna().sum()
)

print(
    "Ship Date :",
    df["Ship Date"].isna().sum()
)
```

## 11. CONTRÔLE DES DÉLAIS DE LIVRAISON

```python
df["Shipping Days"] = (
    df["Ship Date"] -
    df["Order Date"]
).dt.days

print("\n--- Délai de livraison ---")

print(
    df["Shipping Days"]
    .describe()
)
```

## 12. IDENTIFICATION DES VALEURS ANORMALES

```python
print("\nCommandes avec délai négatif :")

print(
    (df["Shipping Days"] < 0).sum()
)


print("\nCommandes avec quantité <= 0 :")

print(
    (df["Quantity"] <= 0).sum()
)


print("\nCommandes avec ventes <= 0 :")

print(
    (df["Sales"] <= 0).sum()
)
```

## 13. RAPPORT FINAL

```python
quality_report = pd.DataFrame({

    "Metric": [
        "Rows",
        "Columns",
        "Duplicates",
        "Missing Values",
        "Negative Sales",
        "Negative Quantity",
        "Invalid Order Dates",
        "Invalid Ship Dates"
    ],

    "Value": [
        len(df),
        len(df.columns),
        duplicates,
        df.isnull().sum().sum(),
        (df["Sales"] < 0).sum(),
        (df["Quantity"] < 0).sum(),
        df["Order Date"].isna().sum(),
        df["Ship Date"].isna().sum()
    ]
})
```

```python
print("QUALITY REPORT")
print(quality_report)
```
